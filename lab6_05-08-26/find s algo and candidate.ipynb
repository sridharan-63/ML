```python
import csv
data = []
with open('Find-S csv.csv', 'r') as csvFile:
    reader = csv.reader(csvFile)
    for row in reader:
        data.append(row)
num_attributes = len(data[0]) - 1
S = ['0'] * num_attributes
G = [['?' for i in range(num_attributes)]]
print("\nTraining Data:\n")
for row in data:
    print(row)
print("\nInitial Specific Hypothesis:")
print(S)
print("\nInitial General Hypothesis:")
print(G)
for i, instance in enumerate(data):
    if instance[-1] == 'Yes': 
        if S == ['0'] * num_attributes:
            S = instance[:-1]
        else:
            for j in range(num_attributes):
                if S[j] != instance[j]:
                    S[j] = '?'
        G = [g for g in G if all(g[k] == '?' or g[k] == S[k] or S[k] == '?' for k in range(num_attributes))]
    else:                        
        G_new = []
        for g in G:
            for j in range(num_attributes):
                if g[j] == '?':

                    if S[j] != '?' and S[j] != instance[j]:
                        new_g = g.copy()
                        new_g[j] = S[j]
                        G_new.append(new_g)
        G = G_new
    print("\nTraining Instance", i + 1)
    print("Specific Hypothesis (S):", S)
    print("General Hypothesis (G):")
    for g in G:
        print(g)
print("\nFinal Specific Hypothesis:")
print(S)
print("\nFinal General Hypothesis:")
for g in G:
    print(g)
```
