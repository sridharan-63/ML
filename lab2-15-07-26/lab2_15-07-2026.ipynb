```python
#lab Ex-3 listing 2
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
iris = load_iris()

df = pd.DataFrame(iris.data, columns=iris.feature_names)

df['variety'] = iris.target

df['variety'] = df['variety'].map({0:'setosa', 1:'versicolor', 2:'virginica'})
data = df.iloc[1:20]

data.plot.scatter(x='sepal length (cm)', y='sepal width (cm)')
plt.title("Scatter Plot: Sepal Length vs Width")
plt.show()

data.hist(column='sepal length (cm)', by='variety')
plt.suptitle("Histogram of Sepal Length by Variety")
plt.show()

data.boxplot(column='sepal length (cm)', by='variety')
plt.title("Boxplot of Sepal Length by Variety")
plt.suptitle("")
plt.show()
```

```python
#lab-4 listing 1
import seaborn as sns
import pandas as pd
from matplotlib import pyplot as plt

data = {
    'week': [1,2,3,4,5],
    'sales':[1.2,1.8,2.6,3.2,3.8],            
    'region':['south','south','north','north','north']
}
df = pd.DataFrame(data)

df1 = sns.load_dataset('iris')

sns.set(color_codes=True)

print("\nBar Plot of week vs Sales\n")
sns.barplot(x="week", y="sales", hue="region", data=df)
plt.title('Bar plot for weeks Vs Sales Data')
plt.show()

print("\nBox Plot of week and sales data\n")
sns.boxplot(data=df[['week','sales']])
plt.xlabel("Variables")
plt.ylabel("Values")
plt.show()

sns.histplot(df1['sepal_width'], kde=False)
plt.title('Histogram of sepal_width')
plt.xlabel("Sepal Width")
plt.ylabel("Frequency")
plt.show()

sns.kdeplot(df1['sepal_width'])
plt.title('Density plot of sepal_width')
plt.xlabel("Sepal Width")
plt.ylabel("Density")
plt.show()

print("\nViolin Plot of week and sales data\n")
sns.violinplot(data=df[['week','sales']])
plt.title('Violin Plot for week and sales data')
plt.xlabel("Variables")
plt.ylabel("Values")
plt.show()

sns.countplot(x="region", data=df, palette="Blues")
plt.title('Count plot based on regions')
plt.xlabel("Region")
plt.ylabel("Count")
plt.show()

print('\nStrip plot\n')
sns.stripplot(x="species", y="petal_length", data=df1)
plt.title('Strip plot for petal length and species')
plt.show()
```

```python
#lab Ex-5 listig 1
from scipy.stats import chi2_contingency
from scipy.stats import chi2
from scipy.stats import ttest_ind

table = [[35, 15], [25, 25]]
print("\nObserved Table:")
print(table)

stat, p, dof, expected = chi2_contingency(table)
print("\nExpected Table:")
print(expected)
prob = 0.90
critical = chi2.ppf(prob, dof)

print("\nCritical Value:", critical)
print("Chi-square Statistic:", stat)

if abs(stat) >= critical:
    print("Stat value >= critical: Reject Null Hypothesis")
else:
    print("Stat value < critical: Accept Null Hypothesis")

x = [6,3,4,5,5,9,7,8]
y = [4,4,1,3,5,6,2,7]
print('\nT-Test completed\n')
stat, p = ttest_ind(x, y, equal_var=True)
print("Statistical value:", stat)
print("p-value:", p)
```
